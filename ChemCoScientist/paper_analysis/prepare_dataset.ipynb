```python
import pandas as pd
```

```python
df = pd.read_csv('questions_dataset_0.2.csv')
df.head()
```

```python
df.rename(columns={"paper_name":"pdf", "Ответ":"answer"}, inplace=True)
```

```python
data = df.copy()
```

```python
data = data[['pdf', 'question', 'answer']]
```

```python
data.head()
```

```python
data = data[data.question.isna()==False]
data.head()
```

```python
data = data[data.pdf!='molecules-28-02420-v2.pdf']
data.shape
```

```python
data
```

```python
data.pdf = data['pdf'].apply(lambda x: x+'.pdf' if x.endswith('pdf')==False else x)
```

```python
data.head()
```

```python
data.pdf.unique()
```

```python
data.to_csv('questions_dataset_0.2_prepared.csv', index=False)
```
